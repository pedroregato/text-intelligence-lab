## 2. Como o TIL está organizado

O TIL separa claramente **fonte** e **execução**:

```text
GitHub
  │  código e histórico oficial
  ▼
Kaggle
  │  execução educacional e experimental
  ▼
Outputs e evidências
```

- **GitHub** é o *source of truth*: guarda a versão canônica dos notebooks e da documentação.
- **Kaggle** é o ambiente principal de execução das aulas e experimentos.
- **Notebook** é o material didático executável que conecta explicação, código, observação e interpretação.

## 4. Inspecionando o ambiente

Execute a célula abaixo. Ela coleta apenas informações básicas do runtime.


````python
import sys
import json
import platform
from pathlib import Path
from datetime import datetime, timezone

is_kaggle = Path('/kaggle').exists()

environment = {
    'python_version': sys.version.split()[0],
    'python_full_version': sys.version,
    'platform': platform.platform(),
    'implementation': platform.python_implementation(),
    'working_directory': str(Path.cwd()),
    'is_kaggle': is_kaggle,
}

environment
````

### O que observar

Procure especialmente estes campos:

- `python_version`: versão real do Python fornecida pelo Kaggle;
- `platform`: sistema operacional do ambiente remoto;
- `working_directory`: diretório em que o notebook está executando;
- `is_kaggle`: deve ser `True` quando a aula estiver sendo executada no Kaggle.

**Checkpoint 1:** se `is_kaggle` for `True`, você acabou de confirmar que o código está sendo executado no ambiente remoto do curso.


## 5. Produzindo uma evidência de execução

Agora vamos transformar a observação anterior em uma pequena evidência reproduzível.

A célula seguinte cria um arquivo JSON com os dados do ambiente e o resultado desta validação.


````python
output_dir = Path('/kaggle/working') if is_kaggle else Path.cwd()

evidence = {
    'experiment_id': 'EDU-INFRA-001',
    'notebook': '00-til-environment-validation.ipynb',
    'executed_at_utc': datetime.now(timezone.utc).isoformat(),
    'runtime': environment,
    'result': 'PASS' if is_kaggle else 'LOCAL_ONLY',
}

evidence_path = output_dir / 'til_environment_evidence.json'
evidence_path.write_text(json.dumps(evidence, indent=2), encoding='utf-8')

print(f'Evidence written to: {evidence_path}')
print(json.dumps(evidence, indent=2))
````

### O que aconteceu aqui?

O notebook fez três coisas simples:

1. reuniu informações do ambiente;
2. registrou quando a execução ocorreu;
3. gravou tudo em `til_environment_evidence.json`.

Quando executado no Kaggle, o arquivo é salvo em `/kaggle/working`, diretório usado para outputs persistidos pela execução.

**Checkpoint 2:** confirme que a saída mostra `"result": "PASS"`.


## 6. Observação → interpretação

No TIL, executar código não é suficiente. O ciclo pedagógico será:

```text

```

Nesta aula, a interpretação é direta:

- `is_kaggle = true` confirma o ambiente de execução;
- `result = PASS` confirma a validação operacional;
- o arquivo JSON fornece uma evidência que pode ser inspecionada posteriormente.

Esse mesmo hábito será reutilizado em experimentos futuros mais interessantes.
