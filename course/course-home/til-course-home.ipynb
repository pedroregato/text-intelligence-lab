# Text Intelligence Lab (TIL)

## *An evidence-driven path to Applied AI Engineering*

**Aprenda a transformar texto em informação, modelos e sistemas capazes de apoiar decisões reais.**

Grande parte da informação produzida pelas organizações não está em tabelas: está em documentos, e-mails, contratos, chamados, avaliações de clientes, transcrições e relatórios. O TIL começa nesse território — **Text Intelligence** — e evolui progressivamente para a engenharia de sistemas modernos de IA.

A jornada segue uma ideia simples:

```text
Text Intelligence
→ Model Engineering
→ Intelligent Orchestration
→ Agentic Systems
```

> **👉 Comece aqui: [Aula 0 — Como o TIL funciona e validação do ambiente](https://www.kaggle.com/code/pedrogentil/til-00-environment-validation)**

O objetivo não é apenas conhecer tecnologias mais avançadas. É aprender **quando uma capacidade adicional realmente melhora uma solução e quando apenas adiciona custo, latência, risco ou complexidade**.

## Que problemas você vai aprender a resolver

### Como classificar textos automaticamente?

**Por que isso importa?** Separar documentos, encaminhar chamados e organizar solicitações consome tempo quando feito manualmente. Um classificador torna esse trabalho rápido, consistente e escalável.

### Como saber se um modelo está realmente funcionando?

**Por que isso importa?** Uma acurácia alta pode esconder erros graves. Precisamos saber que erros o modelo comete, em quais classes falha e se o desempenho é adequado ao impacto da decisão.

### Quando TF-IDF e modelos clássicos bastam — e quando precisamos de embeddings ou Transformers?

**Por que isso importa?** Modelos mais sofisticados não são automaticamente melhores. A escolha certa equilibra qualidade, custo, latência, manutenção e explicabilidade.

### Como fazer um LLM responder com base em documentos confiáveis?

**Por que isso importa?** Um LLM sozinho pode produzir respostas sem sustentação suficiente. Retrieval, grounding e RAG permitem aproximar respostas de evidências recuperadas e verificáveis.

### Como combinar modelos e capacidades diferentes sem transformar o sistema em complexidade desnecessária?

**Por que isso importa?** Sistemas modernos podem envolver routing, LLMs, retrieval, tools e workflows. A engenharia está em escolher a composição mínima que produz valor mensurável.

### Quando vale a pena dar autonomia a um agente?

**Por que isso importa?** Mais autonomia traz novas capacidades, mas também custo, risco e dificuldade de auditoria. Saber quando um agente agrega valor — e quando um workflow determinístico basta — é uma competência central de Applied AI Engineering.

### Como transformar um experimento em uma solução reproduzível?

**Por que isso importa?** Um notebook que funciona uma vez ainda não é uma solução. Projetos reais precisam ser repetíveis, versionados, observáveis e compreensíveis meses depois.

> **Text Intelligence continua sendo a origem do TIL.** Topic Modeling, NER e extração estruturada de informação permanecem no roadmap como extensões formais dessa camada.

## A ideia central do curso

> **Complexidade arquitetural precisa ser conquistada por evidência.**

Ao longo do TIL, uma regra aparece repetidamente:

```text
maior capacidade de modelo ≠ maior utility
maior autonomia           ≠ maior utility
mais planejamento         ≠ maior utility
```

**Utility** é o valor líquido que uma solução entrega no contexto de uso, considerando qualidade, custo, latência, risco e o grau de complexidade ou autonomia necessário para produzir esse resultado.

A pergunta que guia o curso não é:

> *Qual tecnologia é mais avançada?*

Mas sim:

> **Que capacidade adicional melhora o resultado o suficiente para justificar seu custo, latência, risco e complexidade?**

Por isso, o TIL trabalha com outro princípio:

> **Primeiro o mecanismo; depois o framework.**

A proposta é construir competência por meio de experimentos pequenos, observáveis e acumulativos.

### Pré-requisitos

É recomendável ter:

- Python básico;
- noções de pandas;
- fundamentos de Machine Learning supervisionado.

Você **não precisa** começar dominando Transformers, LLMs, RAG ou agentes. A progressão constrói esses conceitos gradualmente.

### Dedicação

Como referência, reserve **cerca de 1–2 horas por aula** para leitura, execução, interpretação e exercícios. Algumas experiências mais pesadas podem exigir tempo adicional.

Os notebooks usam **internet desligada por padrão** sempre que possível. Necessidades específicas de internet, GPU ou recursos adicionais são indicadas na própria aula.

## Ao final desta jornada, você será capaz de

- formular problemas de texto e IA de forma testável e mensurável;
- representar e explorar texto com técnicas adequadas ao problema;
- construir e comparar modelos clássicos, Transformers e sistemas generativos;
- avaliar resultados com métricas, evidências e utility;
- distinguir retrieval, grounding, factuality e governança;
- projetar sistemas com routing, tools, workflows e protocolos de integração;
- projetar sistemas com workflows, protocolos de integração e agentes, evoluindo posteriormente para conceitos como Agent Harness, planning e orchestration runtimes (Aulas 17–22);
- avaliar quando autonomia e planning realmente acrescentam utility;
- transformar experimentos em soluções reproduzíveis e versionadas;
- discutir decisões de arquitetura, custo, risco e trade-offs com clareza.

Em outras palavras, o TIL parte de Text Intelligence e constrói progressivamente o raciocínio necessário para **Applied AI Engineering**.

## Como estudar

Cada aula do TIL possui um notebook oficial publicado no Kaggle.

1. Abra o notebook oficial.
2. Clique em **Copy & Edit** para criar sua cópia.
3. Trabalhe somente na sua cópia.
4. Execute as células em ordem.
5. Observe outputs e checkpoints.
6. Nos exercícios, tente primeiro; use `qN.hint()` se travar e `qN.solution()` para conferir.

```text
Notebook oficial = referência do curso
Cópia do aluno    = ambiente pessoal de aprendizagem
```

Cada aula segue, quando faz sentido:

**Objetivos → Contexto → Ambiente → Experimento → Observação → Interpretação → Exercício → Reprodutibilidade → Resumo**

O ciclo de aprendizagem do laboratório é:

```text
Arquitetar
→ Implementar pequeno
→ Executar
→ Observar
→ Avaliar
→ Corrigir
→ Versionar
→ Expandir
```
