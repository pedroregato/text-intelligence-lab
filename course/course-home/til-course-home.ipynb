# Text Intelligence Lab (TIL) with Kaggle

**Curso prático de Text Intelligence, NLP e Machine Learning com execução no Kaggle e versionamento no GitHub.**

Bem-vindo ao **Text Intelligence Lab (TIL)**.

Este curso foi desenhado para combinar fundamentos, experimentação e prática de engenharia em notebooks executáveis. O objetivo não é apenas ler conceitos, mas aprender por meio do ciclo:

```text
Arquitetar
→ Implementar pequeno
→ Executar
→ Observar
→ Avaliar
→ Corrigir
→ Versionar
→ Expandir
```


## Como usar este curso

Cada aula do TIL possui um notebook oficial publicado no Kaggle.

Antes de começar uma aula:

1. abra o notebook oficial;
2. crie sua própria cópia no Kaggle;
3. trabalhe somente na sua cópia;
4. execute as células na ordem;
5. observe os outputs e checkpoints;
6. use as dicas e soluções apenas quando necessário.

```text
Notebook oficial = referência do curso
Cópia do aluno    = ambiente pessoal de aprendizagem
```

## Padrão das aulas

Os notebooks do TIL seguem, quando fizer sentido, esta estrutura:

1. **Objectives** — o que você deve aprender;
2. **Context** — por que o tema importa;
3. **Environment** — o ambiente necessário;
4. **Experiment** — execução prática;
5. **Observation** — o que aconteceu;
6. **Interpretation** — como entender o resultado;
7. **Exercise** — prática guiada;
8. **Reproducibility** — como repetir a execução;
9. **Summary** — fechamento da aula.

Essa estrutura é uma referência, não uma camisa de força. Se uma seção não agregar valor didático, ela pode ser omitida.

## Reprodutibilidade e qualidade

Uma aula só é considerada pronta quando:

- o notebook executa no Kaggle;
- os outputs esperados são observáveis;
- warnings e erros relevantes foram avaliados;
- o conteúdo passou por revisão pedagógica;
- o notebook canônico está versionado no GitHub;
- o aluno consegue seguir a aula sem depender de instruções externas.
