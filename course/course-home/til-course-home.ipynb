# Text Intelligence Lab (TIL) with Kaggle

**Aprenda a transformar texto em informação, modelos e sistemas capazes de apoiar decisões reais.**

Grande parte da informação produzida pelas organizações não está organizada em tabelas. Ela está em documentos, e-mails, contratos, chamados, avaliações de clientes, transcrições de atendimento, relatórios, notícias, pesquisas e milhares de outros textos.

Transformar esse conteúdo em informação útil é o território da **Text Intelligence**.

Ao longo do **Text Intelligence Lab (TIL)**, você aprenderá progressivamente a enfrentar problemas como estes:

### Qual é o assunto principal de milhares de documentos?

**Por quê isso importa?** Porque organizações acumulam volumes de texto muito maiores do que uma equipe humana consegue ler. Identificar temas, padrões e agrupamentos permite compreender grandes acervos, detectar tendências e priorizar o que merece atenção.

### Como classificar automaticamente textos em categorias?

**Por quê isso importa?** Porque tarefas como separar documentos, encaminhar chamados, organizar solicitações ou identificar tipos de conteúdo consomem tempo quando feitas manualmente. Um classificador pode transformar esse trabalho repetitivo em um processo rápido, consistente e escalável.

### Como identificar intenção, sentimento, entidades e padrões de linguagem?

**Por quê isso importa?** Porque o significado de um texto vai muito além das palavras isoladas. Saber se um cliente está reclamando, qual produto ele menciona, o que pretende fazer ou quais pessoas e organizações aparecem em um documento permite transformar linguagem em sinais úteis para decisão e automação.

### Como descobrir quais termos realmente diferenciam um grupo de textos de outro?

**Por quê isso importa?** Porque nem toda palavra tem o mesmo valor informativo. Identificar quais termos ajudam a distinguir classes, temas ou comportamentos melhora a interpretação dos dados e ajuda a construir modelos mais eficientes e explicáveis.

### Como medir se um modelo de classificação está funcionando bem?

**Por quê isso importa?** Porque uma acurácia aparentemente alta pode esconder erros graves. Em aplicações reais, precisamos saber que tipos de erro o modelo comete, em quais classes ele falha e se seu desempenho é realmente adequado ao impacto da decisão que será automatizada.

### Quando técnicas clássicas como TF-IDF e Naive Bayes são suficientes — e quando precisamos de embeddings ou transformers?

**Por quê isso importa?** Porque modelos mais sofisticados não são automaticamente melhores para todo problema. A escolha correta envolve equilibrar qualidade, custo, latência, volume de dados, facilidade de manutenção e necessidade de explicação.

### Como transformar um experimento de NLP em uma solução reproduzível e tecnicamente confiável?

**Por quê isso importa?** Porque um notebook que funciona uma vez não é ainda uma solução. Projetos reais precisam ser repetíveis, auditáveis, versionados e compreensíveis por outras pessoas — inclusive meses depois da primeira execução.

### Como combinar modelos tradicionais, LLMs e ferramentas modernas sem perder controle sobre custo, qualidade e explicabilidade?

**Por quê isso importa?** Porque sistemas modernos de IA raramente dependem de uma única técnica. O desafio profissional está em escolher onde usar regras, Machine Learning clássico, embeddings, transformers ou LLMs e integrar essas peças de forma econômica, observável e confiável.

O objetivo do curso não é apenas ensinar algoritmos isolados.

Você irá desenvolver uma forma de pensar sobre problemas envolvendo linguagem: **entender o problema, representar o texto, construir uma solução mínima, executar um experimento, observar evidências, medir resultados, identificar limitações e evoluir a solução de maneira disciplinada**.

Por isso, o TIL combina três dimensões que normalmente aparecem separadas:

**Fundamentos**  
Compreender como computadores representam, processam e aprendem a partir de linguagem.

**Experimentação**  
Testar hipóteses em notebooks, comparar abordagens e interpretar resultados com critérios estatísticos e de Machine Learning.

**Engenharia**  
Construir soluções reproduzíveis, versionadas e preparadas para evoluir além do notebook.

A jornada começa pelas técnicas fundamentais de NLP e avança gradualmente para abordagens modernas de representação semântica, transformers e sistemas apoiados por modelos de linguagem.

Ao dominar esse percurso, você terá repertório para atuar em problemas como classificação automática de documentos, análise de voz do cliente e contact centers, triagem de chamados e mensagens, análise de sentimento e intenção, extração de informação, organização e busca em grandes coleções documentais, análise de pesquisas abertas, enriquecimento de bases de conhecimento e aplicações modernas de IA envolvendo NLP e LLMs.

Esse conhecimento é especialmente relevante para profissionais que desejam evoluir em áreas como **Data Science, Machine Learning, NLP, AI Engineering, Analytics, automação inteligente e desenvolvimento de produtos de dados**.

O TIL, entretanto, não promete atalhos.

A proposta é construir competência por meio de experimentos pequenos, observáveis e acumulativos.

Nosso ciclo de aprendizagem será:

```text
Arquitetar
→ Implementar pequeno
→ Executar
→ Observar
→ Avaliar
→ Corrigir
→ Versionar
→ Expandir
```

Cada notebook é, portanto, mais do que uma aula.

É um pequeno laboratório.

E ao final da jornada, o objetivo é que você não apenas conheça técnicas de Text Intelligence — mas tenha desenvolvido o raciocínio necessário para **escolher, construir, avaliar e evoluir soluções reais baseadas em texto**.

**Bem-vindo ao Text Intelligence Lab.**


## Ao final desta jornada, você será capaz de...

- **formular problemas de Text Intelligence** de maneira testável e mensurável;
- **preparar, representar e explorar dados textuais** com técnicas adequadas ao problema;
- **construir e comparar modelos** clássicos, Transformers e sistemas generativos;

- **transformar experimentos em soluções reproduzíveis**, versionadas e preparadas para evolução técnica;

Essas competências formam uma ponte entre análise de dados, Machine Learning, engenharia de IA e sistemas inteligentes baseados em linguagem.

## Onde estamos agora

```text
texto
→ features
→ modelos
→ avaliação
→ routing
→ geração
→ retrieval
→ RAG
→ tools
```

### Política de internet nos notebooks oficiais

Por padrão, os notebooks do TIL executam com **internet desabilitada** para reduzir dependências externas e aumentar a reprodutibilidade.

Quando uma aula precisar intencionalmente de recursos externos — como modelos pré-treinados, APIs ou hubs de modelos — essa dependência será explicitada e justificada.

```text
Internet OFF por padrão → maior controle e reprodutibilidade
Internet ON por exceção → quando o acesso externo fizer parte da aprendizagem
```
