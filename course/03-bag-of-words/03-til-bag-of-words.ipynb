Considere duas mensagens:

```text

```

Nós entendemos o significado das frases. Um modelo tradicional de Machine Learning, porém, normalmente espera receber uma matriz numérica.

Precisamos, então, construir uma ponte:

```text

→ tokens

→ representação numérica

```

Nesta aula vamos parar exatamente na etapa das **features**.


## 3. O que é Bag-of-Words?

Bag-of-Words representa um documento pela ocorrência das palavras de um vocabulário.

A ideia é simples:

1. observar todos os documentos;
2. construir um vocabulário;
3. criar uma coluna para cada termo;
4. contar quantas vezes cada termo aparece em cada documento.

A ordem original das palavras não é representada diretamente.

**Por que o nome “saco de palavras”?** Porque a representação se comporta como se colocássemos as palavras em um saco e observássemos apenas quais aparecem e em que quantidade.


## 4. Construindo nossa primeira matriz documento-termo

Vamos usar a classe `CountVectorizer` da biblioteca `scikit-learn`.

Ela automatiza três etapas importantes:

- identifica o vocabulário;
- transforma cada termo em uma feature;
- produz a matriz de contagens.

Execute a próxima célula.


````python
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "o atendimento foi excelente",
    "não gostei do atendimento",
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

features = vectorizer.get_feature_names_out()

print("Vocabulário:", features)
print("Matriz:\n", X.toarray())
````

### O que observar

Cada **linha** da matriz representa um documento.

Cada **coluna** representa uma palavra do vocabulário.

Cada **valor** indica quantas vezes aquela palavra aparece naquele documento.

Note também algo importante: `CountVectorizer` possui decisões próprias de tokenização. Por padrão, tokens muito curtos podem não entrar no vocabulário. Por isso, a palavra isolada `o` não aparece como feature.

**Checkpoint 1:** confirme que você consegue associar cada posição da matriz a um termo do vetor `features`.


## 5. Tornando a matriz legível

Uma matriz numérica é útil para algoritmos, mas não é a forma mais amigável para inspeção humana.

Vamos transformá-la temporariamente em um `DataFrame` para enxergar claramente as features.


````python
import pandas as pd

bow_df = pd.DataFrame(
    X.toarray(),
    columns=features,
    index=["documento_1", "documento_2"],
)

bow_df
````

### Onde estão as features?

Agora elas estão visíveis: **cada coluna é uma feature**.

Por exemplo, a coluna `excelente` informa se e quantas vezes esse termo apareceu em cada documento.

Essa é uma mudança conceitual importante:

```text
palavra
→ coluna numérica
→ feature
```

O texto ainda existe como linguagem para nós, mas o algoritmo passa a enxergá-lo como um vetor de números.

## 7. Uma limitação importante: ordem das palavras

Compare:

```text
cliente não gostou
cliente gostou não
```

Um Bag-of-Words simples pode produzir contagens muito semelhantes, embora a organização linguística seja diferente.

Isso acontece porque a representação básica registra presença e frequência, mas não preserva diretamente a sequência completa das palavras.

**Por que isso importa?** Porque representação é sempre uma escolha: ganhamos simplicidade e transparência, mas perdemos parte da estrutura linguística.


## 8. Outro detalhe importante: matriz esparsa

Em um corpus real podemos ter milhares ou milhões de termos, mas cada documento utiliza apenas uma pequena parte deles.

Por isso, `CountVectorizer` retorna por padrão uma **matriz esparsa** (*sparse matrix*), que armazena de forma eficiente principalmente os valores diferentes de zero.

Nesta aula usamos `.toarray()` apenas porque nosso exemplo é minúsculo e queremos enxergar a matriz.

> Em datasets grandes, converter toda a matriz para um array denso pode consumir muita memória.
